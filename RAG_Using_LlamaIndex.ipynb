```python
from llama_index.core import SimpleDirectoryReader
from llama_index.core.node_parser import SentenceSplitter, TokenTextSplitter
```

```python
documents = SimpleDirectoryReader(input_files=["transformer.txt"]).load_data()
documents
```

```python
len(documents)
```

```python
splitter = SentenceSplitter(chunk_size=1024, chunk_overlap=20)
```

```python
nodes = splitter.get_nodes_from_documents(documents)
```

```python
len(nodes)
```

```python
nodes[0].text
```

```python
import tiktoken
for i, node in enumerate(nodes):
  encoding = tiktoken.encoding_for_model("gpt-3.5-turbo")
  print(f"number of tokens in node {i+1}: {len(encoding.encode(node.text))}")
```

```python
from llama_index.llms.openai import OpenAI
import os
from llama_index.core.llms import ChatMessage
```

```python
# Fix: add base_url to point to OpenRouter
llm = OpenAI(
    model="gpt-3.5-turbo",                        # Fix: typo corrected
    api_key="Your key",     # Fix: use your OpenRouter key
    api_base="https://openrouter.ai/api/v1"        # Fix: OpenRouter endpoint
)
```

```python
messages = [
    ChatMessage(role = 'system', content = 'answer the question accurately'),
    ChatMessage(role = 'user', content = 'what is the capital of france')
    ]
```

```python
response = llm.chat(messages)
print(response.message.content)
```
