# Chunking Straregy

```python
def chunk_text(text, source):
    """
    Split a document into chunks by paragraph.
    Returns a list of dicts with text and metadata.
    """
    paragraphs = text.strip().split("\n\n")

    chunks = []
    for para in paragraphs:
        para = para.strip()
        # Skip short lines (headers, separators)
        if len(para) < 50:
            continue
        # Skip separator lines
        if para.startswith("===="):
            continue
        chunks.append({
            "text": para,
            "source": source
        })

    return chunks

```

```python
hr_chunks = chunk_text(hr_document, "HR Policy")
engineering_chunks = chunk_text(engineering_document, "Engineering")
onboarding = chunk_text(onboarding_document, "Onboarding")
product = chunk_text(product_document, "Product")
security = chunk_text(security_document, "Security")
```

```python
all_chunks = hr_chunks + engineering_chunks + onboarding + product + security
print(len(all_chunks))
```

# Manual Embedding

```python
from sentence_transformers import SentenceTransformer

model = SentenceTransformer("sentence-transformers/all-MiniLM-L6-v2")

sentences = ''' Office Network:
- Connect only to the "NovaTech-Secure" WiFi network (WPA3)
- The "NovaTech-Guest" network is for visitors only — employees should not use it
- Do not connect personal IoT devices (smart speakers, etc.) to the office network '''

embeddings = model.encode(sentences)
print(embeddings.shape)
print(embeddings)
```

```python
sentence1 = ''' I love chicken Biriyani '''
sentence2 = ''' I love eating flaovured rice with chicken'''
sentence3 = ''' I love mutton Biriyani'''

embeddings1 = model.encode(sentence1)
embeddings2 = model.encode(sentence2)
embeddings3 = model.encode(sentence3)

print(embeddings1.shape)
print(embeddings2.shape)
print(embeddings3.shape)
```

```python
from sklearn.metrics.pairwise import cosine_similarity

similarity_score1 = cosine_similarity([embeddings1], [embeddings2])
similarity_score2 = cosine_similarity([embeddings1], [embeddings3])
similarity_score3 = cosine_similarity([embeddings2], [embeddings3])

print(similarity_score1)
print(similarity_score2)
print(similarity_score3)
```

# Storing Chunks with Chroma DB

```python
chroma_client = chromadb.Client()
collection = chroma_client.create_collection(name="documents")
```

```python
documents = []
ids = []
metadatas = []

for i, chunk in enumerate(all_chunks):
    documents.append(chunk["text"])
    ids.append(f"chunk_{i}")
    metadatas.append({"source": chunk["source"]})

print(documents[0])
print(ids[0])
print(metadatas[0])
```

```python
collection.add(
    documents=documents,
    ids=ids,
    metadatas=metadatas
)
```

```python
print(f" Stored {len(documents)} chunks in chroma db")
print(f" Sources: 5 differnt files")
```

# Building Retrievel Pipeline

```python
def retrive(question, n_results=3):
    results = collection.query(
        query_texts= [question],
        n_results= n_results
        )
    return results["documents"][0], results["metadatas"][0]
```

```python
chunks, sources = retrive("what is the work from home policy ?", 10)
#print(retrive("what is the work from home policy ?", 3))
for i in range(len(chunks)):
    print(f"--- Chunks {i+1} ---")
    print(f"Source : {sources[i]}")
    print(f"Text : {chunks[i]}")
    print("\n\n")
```
